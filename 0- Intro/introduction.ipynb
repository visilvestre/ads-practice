**Algorithm**: Set of instructions to accomplish a task

##### Arrays,  Linked Lists, Stacks, and Hash Tables:
- **Linked List**: All blocks of memory point to each other, but to go through all of them, it is necessary to start at the first and go 1 by 1 until you reach the desired item. [use pointers]
<br>(O(1) insertion, O(1) deletion, O(n) reading)

  - **Pointer**: Use a little bit of memory at each position to store the address of the next item.
  - Can only do sequential access
  - Can't read batches of numbers at once
  - Use more memory given the need to save the pointer to the next memory slot

- **Arrays**: You know the address for all of your elements in the array, but insertion takes more time. 
<br> (O(n) insertion, O(n) deletion, O(1) reading)
    - PS: Arrays are often more used given they have advantages over linked lists.
        - Arrays are better at reads
        - Can do sequential access (read one element at a time in sequence)
            - Can do faster batch sequential reads, given the values are nearby
        - Can do random access (large volume of use cases)

- **Stack**: Items inserted to a stack are added to the top (Push), and items read and deleted from the stack are taken from the top (pop). Recursion take advantage of the stack the computer created when running codes.

- **Hash Tables**: Items are stored in pairs ("string":number) (aka "Keys":Values), and the hash function becomes responsible to find the string in O(1) and return the number. (Also known as Injective Function). In python the hash tables are called dictionaries.
    - Can suffer with collisions if more than 1 item is stored at the same place 
        - In case of collisions, sometimes the creation of a linked list within the collision space help without augmenting too much the retrieval cost.
    - Use Cases: Good for lookups, DNS, preventing duplicate entries, cache
    - **Load Factor** of a hash table:
        - $ \dfrac{Number items in hash table}{Total number of slots} $
    - Good hash functions distribute items evenly within the array, and have low load factors (<0.7)
    - Curiosity: Google Abseil library uses "CityHash" as a function
