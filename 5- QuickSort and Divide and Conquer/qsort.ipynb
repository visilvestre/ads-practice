## Divide and Conquer
![""](divideconquer.png "Divide and Conquer - Image by geekforgeeks")

Divide and conquer is a recursive algorithm. there are 2 core steps:

1. Find out the base case, this should be the simplest case possible.
2. Divide the problem until it become the base case. 

This is not a simple algorithm that you can apply, it is instead a way to think about a problem. 

```python
""" Example: Using divide and conquer to sum an array of numbers """
import numpy.random as rand
def divide_and_conquer_sum(arr):
    # Exit case 
    if len(arr) == 0:
        return 0
    
    if len(arr) == 1:
        return arr[0]
    # Divide
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]
    
    # Conquer (recursively sum the halves)
    left_sum = divide_and_conquer_sum(left_half)
    right_sum = divide_and_conquer_sum(right_half)
    
    # Combine
    return left_sum + right_sum

# Create a random array of integers
array_size = 10
random_array = rand.randint(1, 100, size=array_size)
print("Array:", random_array)
total_sum = divide_and_conquer_sum(random_array)
print("Sum of array elements:", total_sum)

print(sum(random_array))
```

#### QuickSort Algorithm
![](qsort.png "Quicksort image by GeekforGeeks")

1. Check if the algorithm have more than 1 item,  if not return the array.

2. select a pivot , then create a list of items that are smaller than the pivot (less) , and one list of items that are greater than the pivot (greater). 

3. recursively call quicksort with the less list on the left, plus the pivot, plus the quicksort with the greater list. 

This will divide and conquer until reaching the minimum elements that will get ordered by the minimum to one side and maximum to the other. 

QuickSort works at $O(n*log{n})$ if the optimal pivots are selected, otherwise it can work up to $O(n^2)$ time

```python
def quicksort(arr):
    if len(arr) < 2:
        return arr
    mid = len(arr) // 2
    pivot = arr[mid]
    less = [x for x in arr if x < pivot]
    greater = [x for x in arr if x > pivot]
    
    return quicksort(less) + [pivot] + quicksort(greater)

unsorted_array = rand.randint(1, 100, size=10)
print("Unsorted Array:", unsorted_array)
sorted_array = quicksort(unsorted_array.tolist())
print("Sorted Array:", sorted_array)
```
