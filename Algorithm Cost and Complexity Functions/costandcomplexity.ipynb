#### Measurement of an algorithm's complexity
1. Analysis of a **specific algorithm**:
Consider the cost to use the algorithm in terms of time and space as a function of the size of the input data.

2. Analysis of a **class of algorithms**:
What is the algorithm with the best performance for a given problem? 
All algorithms that solve the same problem can be compared in terms of time and space complexity.

The **most efficient algorithm  of a class** gives 
the **measure of the complexity** of the problem itself.

An algorithm with the **best performance** is called an **optimal algorithm**. (more than one algorithm can be optimal for the same problem)

#### Complexity Function
The cost function OR complexity function f()

- **Time complexity function:**
f(n) measure the time needed to execute the algorithm for a problem with size n.

- **Space complexity function:**
f(n) measure the memory needed to execute the algorithm for a problem of size n.

f() represents the number of times a given operation that is relevant is executed.







````python
""" Example: Largest Element """
def largest_element(arr):
    max = arr[0]
    n = len(arr)
    for i in range(1,n):
        if max < arr[i]:
            max = arr[i]
    return max
print(f"Largest element is: {largest_element([1, 3, 5, 7, 9, 2, 4, 6, 8])}")
````

Given f() as a complexity function, the number of comparisons executed by the algorithm for the problem is the number of elements of the array arr.

#### The data input size
The execution cost of an algorithm depends primarily on the size of the input data. 

- **Best Case:** Minimum cost of execution for any instance of size n.

- **Worst Case:** Maximum cost of execution for any instance of size n.

- **Average Case (Expected Case):** Average cost of execution for all instances of size n.

PS: It is common to consider the time of execution as a function of the size of the input.



````python
""" Example: Search for a registry """
def search_registry(registry, target):
    """
    Given a key `target`, search for it 
    in the `registry`. 

    A sequential search is performed

    Args:
        registry (list): List to be searched 
        target (): Key to be found

    Returns:
        int: Index of the target in the registry, 
             -1 if not found
    """

    for i in range(1,len(registry)):
        if registry[i] == target:
            return i
    return -1

registry = [6,7,8,9,0,1,2,3,4,5]
target = 3

print(f"Identify the position of the key {target} in the registry.")
print(f"The key is in the index {search_registry(registry, target)}")
````

For the code above: 
- Best Case: $$ f(n) = 1 $$ when the element is the first of the array

- Worst Case: $$ f(n) = n $$ when the element is the last of the array

- Average Case: $$ f(n) = \dfrac{n+1}{2}$$

The average case consider that the element is equally likely to be in any position of the array. 

$$ f(n) = 1*p{1} + 2*p{2} + ... + n*p{n} $$

with that we have: 

$$f(n) = \dfrac{1}{n} (1 + 2 + ... + n) = \dfrac{n+1}{2} $$

#### Asymptotic Dominance
When analyzing the complexity of an algorithm, we are interested in the behavior of the complexity function for large values of n.
For large values of n, lower order terms and constant factors become insignificant.
For example, consider the function:
$$ f(n) = 3n^2 + 5n + 2 $$
For large values of n, the term 3n^2 dominates the function, and we can say that f(n) is asymptotically dominated by n^2.
We denote this as:
$$ f(n) = O(n^2) $$
This notation means that there exists a constant C and a value n0 such that for all n >= n0, f(n) <= C * n^2.

![Asymptotic Dominance](graph1.png "Asymptotic Dominance")

**The asymptotic behavior of f(n) represent the limit of the cost behavior, when n approaches infinity.**

There are 3 common notations used to describe the asymptotic behavior of functions:
1. Big O Notation (O): Describes an upper bound on the growth rate of a function.

![Big O Notation](bigo.png "Big O Notation")

2. Omega Notation (Ω): Describes a lower bound on the growth rate of a function.

![Omega Notation](omega.png "Omega Notation")

3. Theta Notation (Θ): Describes a tight bound on the growth rate of a function.  

![Theta Notation](theta.png "Theta Notation")

#### Comparing programs

We can compare programs based on their complexity functions, neglecting constant factors and lower-order terms.

Hierarchy of common complexity classes from fastest to slowest growth rates:
1. Constant Time: O(1)
2. Logarithmic Time: O(log n)
3. Linear Time: O(n)
4. Linearithmic Time: O(n log n)
5. Quadratic Time: O(n^2)
6. Cubic Time: O(n^3)
7. Exponential Time: O(2^n)
