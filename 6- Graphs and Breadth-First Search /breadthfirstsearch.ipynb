#### Graphs

![](graph.png "Graph representations by www.educba.com")
<br>Graphs are an abstract data structure that models a set of connections. Graphs are made up of nodes and edges. A node can be directly connected to many other nodes. 

How to implement:
- Using Hash: 
    ```python
        graph = {}
        graph["Marta"] = ["John", "Bob" , "Atlas"]
    ```
    - Once a call is made to "Marta" , a list of all neighbors will be retrieved.
    - Sample of a bigger graph:
        ```python
            graph = {}
            graph["Person1"] = ["Person2","Person3"]
            graph["Person2"] = ["Person3"]
            graph["Person3"] = ["Person1","Person2"]
            graph["Person4"] = []
        ```

Types:
- **Directed Graph**: The relationships in the graph are only one way.
    - Nodes can be in-neighbors or out-neighbors depending on where the graph points towards.
    - **Tree** : Special class of graphs where no edges ever point back.
        - **Binary Tree**: Special case of trees where each node have at most 2 children.
- **Undirected Graph**: The relationships does not have any arrows.
    - In undirected graphs, the neighbor don't have in- or out- classifications.



When using queues for graphs in python: enqueue is append, and dequeue is popleft.

#### Breadth-First Search
![](bfs.png "Breadth-First Search image by GeekforGeeks")

Use: Allows to find the shortest distance between 2 things. Also used for spellchecker, or search engine crawlers.

Two steps:
1. Model the problem as a graph
2. Solve the problem using BFS

Helps answer questions such as:
1. Is there a path from node A to node B?
2. What is the shortest path from node A to node B?

Degrees of connection:
<ul>
    <li>1st - Directly connected to the node
    <li>2nd - Connected to 1st degree connections, but not directly with the node
    <li>Nth th - Connected to the (n-1)th degree connections, but not directly with the node or any of it's (n-2)th connections or less.
</ul>

You should not search at the 2nd degree if you did not exhausted the 1st degree. 

BFS runs with O(number of edges + number of people) in our example, commonly written as O(V+E) <br>[V- Vertices, E - Edges]

````python
from collections import deque
# Example from book grokking algorithms

# Define the graph using a dictionary
graph = {} 
graph["you"] = ["alice", "bob", "claire"]
graph["bob"] = ["anuj", "peggy"]
graph["alice"] = ["peggy"]
graph["claire"] = ["thom", "jonny"]
graph["anuj"] = []
graph["peggy"] = []
graph["thom"] = []
graph["jonny"] = []

def person_is_seller(name):
    return name[-1] == 'm'

# Breadth-First Search (BFS) implementation
def bfs(graph, start_node="you"):
    search_queue = deque()
    search_queue += graph[start_node]

    while search_queue:
        person = search_queue.popleft()
        print(f"Checking {person}...")
        if person_is_seller(person):
            print(f"{person} is a mango seller!")
            return True
        else:
            search_queue += graph[person]
    return False

print(f"It is {bfs(graph)} that there is a mango seller in the network.")
````

The case above shows that we can have a same node being visited more than once. To solve that we add a set for the people already visited.

````python
from collections import deque
# Example from book grokking algorithms

# Define the graph using a dictionary
graph = {} 
graph["you"] = ["alice", "bob", "claire"]
graph["bob"] = ["anuj", "peggy"]
graph["alice"] = ["peggy"]
graph["claire"] = ["thom", "jonny"]
graph["anuj"] = []
graph["peggy"] = []
graph["thom"] = []
graph["jonny"] = []

def person_is_seller(name):
    return name[-1] == 'm'

# Breadth-First Search (BFS) implementation
def bfs(graph, start_node="you"):
    search_queue = deque()
    search_queue += graph[start_node]
    searched = set()
    while search_queue:
        person = search_queue.popleft()
        if person not in searched:
            print(f"Checking {person}...")
            if person_is_seller(person):
                print(f"{person} is a mango seller!")
                return True
            else:
                searched.add(person)
                search_queue += graph[person]
    return False

print(f"It is {bfs(graph)} that there is a mango seller in the network.")
````
