#### Tortoise and Hare Algorithm

The Tortoise and Hare algorithm, also known as Floyd's Cycle-Finding Algorithm, is a pointer algorithm that uses two pointers moving at different speeds to detect cycles in a sequence of values. It is commonly used in linked lists to determine if there is a cycle (loop) present.

Finding whether a cycle exists (<a href="https://medium.com/@preetiwrites1111/floyds-tortoise-hare-cycle-finding-algorithm-explanation-proof-bf59befb153a">source</a>):

![](image.png "Tortoise and Hare Algorithm")

Step 1: Initialize a fast(hare) and slow(tortoise) pointer at the head.

Step 2: Traverse the list such that the slow(tortoise) pointer will move one node at a time whereas the fast(hare) pointer will move at twice its speed i.e., two nodes at a time

Step 3: While traversing, if slow(tortoise) and fast(hare) pointers meet at the same node that means cycle exists.

Step 4: If fast(hare) reaches the end(null) that means there is no cycle.





```python
def has_cycle(head):
    slow = head  # Tortoise
    fast = head  # Hare
    while fast and fast.next:
        slow = slow.next          # Move slow pointer by 1
        fast = fast.next.next     # Move fast pointer by 2
        if slow == fast:          # Cycle detected
            return True
    return False  # No cycle detected

# Example usage
class ListNode:
    def __init__(self, value=0, next=None):
        self.value = value
        self.next = next
        
# Creating a linked list with a cycle for testing
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
head.next.next.next = head.next  # Creates a cycle for testing

print(has_cycle(head))  # Output: True
```
