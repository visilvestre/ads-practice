### Two Pointers

![](twopointers.png "Two Pointers")

<BR> The idea of the two pointers technique is to use two different indices to traverse the data structure (usually an array or a string) instead of using a single index. This approach can help optimize the time complexity of certain algorithms, especially when dealing with sorted arrays or strings.

The array must be sorted for the two pointers technique to work effectively.

#### Common Scenarios for Using Two Pointers
1. **Finding Pairs with a Given Sum**: Given a sorted array, find two numbers that add up to a specific target sum.
2. **Removing Duplicates**: Remove duplicates from a sorted array in-place.
3. **Merging Two Sorted Arrays**: Merge two sorted arrays into one sorted array.
4. **Reversing a String**: Reverse the characters in a string.
5. **Partitioning an Array**: Partition an array around a pivot value.

```python
import numpy.random as rand
# Example: Finding Pairs with a Given Sum
def two_sum(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return (arr[left], arr[right])
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return None
# Example usage

# Random integer array generation
arr = rand.randint(1, 10, size=10)
arr.sort()  # Sorting the array for two-pointer technique
print("Array:", arr)
target = arr[2] + arr[4]  # Example target sum
result = two_sum(arr, target)
print(result)  # Output: (2, 4)


# Example: Finding Triplets with a Given Sum
def three_sum(arr, target):
    arr.sort()
    for i in range(len(arr) - 2):
        left, right = i + 1, len(arr) - 1
        while left < right:
            current_sum = arr[i] + arr[left] + arr[right]
            if current_sum == target:
                return (arr[i], arr[left], arr[right])
            elif current_sum < target:
                left += 1
            else:
                right -= 1
    return None

result2 = three_sum(arr, target)
print(result2)  # Output: (1, 2, 7)
```
