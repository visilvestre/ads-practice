#### Selection Sort

![](image.png "Selection Sort Algorithm")


Go through the list searching for the minimum value, and add that to the new list, keep doing this until the new list is fulfilled.

Selection Sort runs in $O(n^2)$

```python
import numpy.random as rand

def find_smallest(arr):
    min = arr[0]
    min_index = 0
    for i in range(1, len(arr)):
        if arr[i] < min:
            min = arr[i]
            min_index = i
    return min_index

def selection_sort(arr):
    new_arr = []
    copied_arr = arr.copy()
    
    for i in range(0, len(arr)-1):
        min_index = find_smallest(copied_arr)
        new_arr.append(copied_arr.pop(min_index))
    return new_arr


# Sample usage
array = rand.randint(0, 100, size=10).tolist()

# transform the array into integer values between 0 and 100
array = list(map(lambda x: int(x), array))

print("Original array:", array)
print(f"Sorted Array: {selection_sort(array)}")  # Output: [2, 3, 5, 6, 10]
```
