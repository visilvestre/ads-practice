# Binary Search

!["Binary Search"](binary.png "Binary Search - Image by geekforgeeks")

The **binary search** is an algorithm that takes a sorted list of elements, and returns an element that being searched. If the element is not in the list, it returns null.


```python
import numpy.random as rand

def binary_search(arr, target):
    """
    Search for a given value in a sorted array using binary search.
    
    Args:
        arr (list): List of sorted elements.
        target (integer): The value to search for.
    
    Returns:
        integer: The index of the target in the array if found, otherwise None.
    """
    
    low, high = 0, len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        guess = arr[mid]
        if guess == target:
            return mid
        elif guess > target:
            high = mid - 1
        else:
            low = mid + 1
    return None
        

sample_list = rand.random(10).tolist()
sample_list = list(map(lambda x: int(x * 100), sample_list))  # Scale to integers for easier searching
sample_list.sort()
print("Sorted List:", sample_list)
target_value = sample_list[rand.randint(0, 10)]  # Pick an arbitrary value from the list to search

result = binary_search(sample_list, target_value)
if result is not None:
    print(f"Value {target_value} found at index {result}.")
else:
    print(f"Value {target_value} not found in the list.")
```

The complexity of the binary search is O(log(n)). Compared to a greedy simple search checking item by item O(n) , it is much faster.
