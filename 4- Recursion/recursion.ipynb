# Recursion

![Recursion](recursion.png "Optional title text")

**When to use recursion**: When the instance of a problem contains smaller instances of the same problem.

**How**: 
- If the problem can be solved directly (base case), do so.
- Otherwise, break the problem down into smaller instances of the same problem (recursive case) and combine the results.

**PS:** A recursive function is not necessarily a function that calls itself. It can also call other functions that eventually lead to the original function being called again (indirect recursion).

```python
""" Exercise 1: Calculating the factorial of a number using recursion. """
def factorial(n):
    # Base case: factorial of 0 or 1 is 1
    if n == 0 or n == 1:
        return 1
    # Recursive case: n! = n * (n-1)!
    else:
        return n * factorial(n - 1)

print(f"The factorial of 5 is: {factorial(5)}")
```

```python
""" Exercise 2: Calculate the number of binaries of number n using recursion. """
def bin(n = int, arr = []):
    """
    Calculate the number of binary digits in integer n.
    Args:
        n (int): The integer to calculate binary digits for.
    Returns:
        int: The number of binary digits in n.
    """
    if (n//2) in (0,1):
        if n==2:
            arr.append(1)
            arr.append(1)
        if n==1:
            arr.append(1)
        elif n==0:
            arr.append(0)
        return arr
    else:
        print(n//2)
        arr.append(n%2)
        return bin(n//2, arr)
    
print(bin(138))

# The complexity of this algorithm is O(log n) 
# because with each recursive call, the input size n is halved (n//2), 
# leading to a logarithmic number of calls relative to the original input size.
```

```python
""" Exercise 3: Q function"""
def Q(n):
    """
    Calculate the Q function recursively.
    Args:
        n (int): The integer to calculate the Q function for.
    Returns:
        int: The result of the Q function.
    """
    if n == 1:
        return 1
    else:
        return Q(n-1) + 2*n - 1

print(Q(5))
# The complexity of this algorithm is O(n)
# The value the function Q calculates is the square of n (n^2).

# 1 + (2*2-1) + (2*3-1) + (2*4 - 1) + (2*5 - 1)
# = 1 + 3 + 5 + 7 + 9 = 25 = 5^2
```

```python
""" Exercise 4: Function M """
def M(v = [], n = int):
    """
    Calculate the M function recursively.
    Args:
        v (list): A list of integers.
        n (int): The number of elements in the list to consider.
    Returns:
        int: The result of the M function.
    """
    if n == 1:
        return v[0]
    else:
        x = int
        x = M(v, n-1)
        if x <= v[n-1]:
            return x
        else:
            return v[n-1]

v = [3, 5, 2, 8, 1, 4]
n = len(v)
print(M(v, n))
# The complexity of this algorithm is O(n)
# The value the function M calculates is the minimum value in the list v.

def Miter(v = [] , n = int):
    """
    Calculate the M function iteratively.
    Args:
        v (list): A list of integers.
        n (int): The number of elements in the list to consider.
    Returns:
        int: The result of the M function.
    """
    min_value = v[0]
    for i in range(1, n):
        if v[i] < min_value:
            min_value = v[i]
    return min_value
print(Miter(v, n))
# The complexity of this algorithm is O(n)
# The value the function Miter calculates is the minimum value in the list v.
```

```python
""" Exercise 5: Fibonacci sequence """
def fibonacci(n):
    """
    Calculate the nth Fibonacci number recursively.
    Args:
        n (int): The position in the Fibonacci sequence.
    Returns:
        int: The nth Fibonnaci number.
    """

    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
print(fibonacci(6))
# The complexity of this algorithm is O(2^n)
# The value the function fibonacci calculates is the nth Fibonacci number.
```
