#### Sliding Window Algorthm
![](slidingwindow.png "Sliding Window Technique")
The sliding window technique is used to solve problems that involve finding a subarray or substring that meets certain criteria. It involves maintaining a window (a subset of the array or string) and adjusting its size and position as you iterate through the data structure.

Here's a simple example of the sliding window technique to find the maximum sum of a subarray of size `k` in an array:

(Time Complexity: O(n), Space Complexity: O(1))


```python
def max_sum_subarray(arr, k):
    """
    Find the maximum sum of a subarray of size k.
    
    Args:

    arr (List[int])  Input array of integers.
    k (int) Size of the subarray.

    Returns

    max_sum (int) Maximum sum of the subarray of size k.
    """
    n = len(arr)
    if n < k:
        return None  # Not enough elements for the window size
    max_sum = float('-inf') # Initialize max sum with negative infinity
    window_sum = sum(arr[:k])  # Initial window sum
    max_sum = window_sum # Update max sum
    for i in range(k, n):
        window_sum += arr[i] - arr[i - k]  # Slide the window
        max_sum = max(max_sum, window_sum)

    return max_sum
# Example usage
arr = [1, 2, 3, 4, 5, 6]
k = 3
result = max_sum_subarray(arr, k)
print(result)  # Output: 15 (4 + 5 + 6)
```
